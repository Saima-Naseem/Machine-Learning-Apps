```python
#import libraries
import pandas as pd
```

```python
# upload files
amazon = pd.read_csv("/content/amazon_cells_labelled.txt", sep="\t", header=None)
imdb = pd.read_csv("/content/imdb_labelled.txt", sep="\t", header=None)
yelp = pd.read_csv("/content/yelp_labelled.txt", sep="\t", header=None)
```

```python
# inspect data
amazon.head()
imdb.head()
yelp.head()
```

```python
# name data columns

amazon.columns = ["review", "sentiment"]
imdb.columns = ["review", "sentiment"]
yelp.columns = ["review", "sentiment"]
amazon.head()
imdb.head()
yelp.head()
```

```python
# Combine all three data sets

df = pd.concat([amazon, imdb, yelp], ignore_index=True)
```

```python
# check data shape

df.shape
```

```python
# check missing values

df.isnull().sum()
```

```python
# check how many negative or positive values are in the dataset

df["sentiment"].value_counts()
```

```python
# import train test split from sklearn

from sklearn.model_selection import train_test_split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df["review"],
    df["sentiment"],
    test_size=0.2,
    random_state=42,
    stratify=df["sentiment"]
)
```

```python
# convert text into numbers

from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)
```

```python
# train the logistic regression model

from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(X_train_tfidf, y_train)
```

```python
# check model accuracy

from sklearn.metrics import accuracy_score

y_pred = model.predict(X_test_tfidf)
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
# check precision, recall F1 score

from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```

```python
# save model

import joblib

joblib.dump(model, "sentiment_model.pkl")
joblib.dump(vectorizer, "tfidf_vectorizer.pkl")
```

```python
# real time model testing - a positive review

text = ["I absolutely loved this product. It was amazing!"]

text_tfidf = vectorizer.transform(text)
prediction = model.predict(text_tfidf)
if prediction[0] == 1:
    print("Positive 😊")
else:
    print("Negative 😞")
```

```python
# real time model testing - a negative review

text = ["This product is terrible. I am very disappointed."]

text_tfidf = vectorizer.transform(text)
prediction = model.predict(text_tfidf)
if prediction[0] == 1:
    print("Positive 😊")
else:
    print("Negative 😞")
```
